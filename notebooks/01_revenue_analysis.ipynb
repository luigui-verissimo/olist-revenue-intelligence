# Revenue Analysis | Análise de Receita

### 🇧🇷 Português

O objetivo desta análise é entender como a receita do marketplace evoluiu ao longo do tempo.

Perguntas principais:

- Como a receita mensal evoluiu?
- Existem padrões de crescimento?
- Qual foi o melhor período de performance?

---

### 🇺🇸 English

The objective of this analysis is to understand how marketplace revenue evolved over time.

Main questions:

- How did monthly revenue evolve?
- Are there growth patterns?
- What was the best-performing period?

```python
from pathlib import Path
import sys

import pandas as pd

project_root = Path.cwd().parent

if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from src.data_ingestion import build_marketplace_base

df = build_marketplace_base()

df.shape
```

```python
df["order_purchase_timestamp"] = pd.to_datetime(
    df["order_purchase_timestamp"]
)

df["month"] = (
    df["order_purchase_timestamp"]
    .dt.to_period("M")
)

monthly_revenue = (
    df.groupby("month")["payment_value"]
    .sum()
)

monthly_revenue.head()
```

```python
best_month = monthly_revenue.idxmax()
best_revenue = monthly_revenue.max()

print("Best Month:", best_month)
print("Revenue:", round(best_revenue, 2))
```
