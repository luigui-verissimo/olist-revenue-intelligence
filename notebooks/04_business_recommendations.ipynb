Objectives:

- Measure commercial risk;
- Evaluate seller concentration;

```python
df = build_marketplace_base()

df.shape
```

```python
seller_revenue = (
    df.groupby("seller_id")["payment_value"]
    .sum()
    .sort_values(ascending=False)
)

seller_revenue_df = (
    seller_revenue
    .reset_index()
)

seller_revenue_df.columns = [
    "seller_id",
    "revenue"
]
```

```python
seller_revenue_df["segment"] = pd.qcut(
    seller_revenue_df["revenue"],
    q=4,
    labels=[
        "Long Tail",
        "Emerging",
        "Strong",
        "Top Performers"
    ]
)

seller_revenue_df.head()
```

```python
seller_revenue_df.columns
```

# Business Recommendations | Recomendações de Negócio

## Revenue Risk Matrix | Matriz de Risco de Receita

### 🇧🇷 Português

Após identificar concentração de receita entre sellers, o próximo passo é medir o risco comercial do marketplace.

Pergunta central:

Quanto da receita está concentrada nos sellers de alta performance?

Objetivo:

Avaliar dependência comercial e exposição a churn estratégico.

---

### 🇺🇸 English

After identifying revenue concentration across sellers, the next step is to measure marketplace commercial risk.

Core question:

How much revenue is concentrated among top-performing sellers?

Objective:

Evaluate commercial dependency and exposure to strategic churn.

```python
segment_revenue_pct = (
    seller_revenue_df
    .groupby("segment")["revenue"]
    .sum()
)

segment_revenue_pct = (
    segment_revenue_pct
    / segment_revenue_pct.sum()
) * 100

segment_revenue_pct = (
    segment_revenue_pct
    .round(2)
)

segment_revenue_pct
```

## Revenue Risk Matrix | Matriz de Risco de Receita

### 🇧🇷 Português

A análise revelou forte concentração comercial entre sellers.

Distribuição da receita:

- **Long Tail:** 0.57%
- **Emerging:** 2.70%
- **Strong:** 9.93%
- **Top Performers:** 86.79%

### Insight estratégico:

O marketplace apresenta alta dependência estrutural dos sellers Top Performers.

### Riscos identificados:

- Churn de sellers estratégicos;
- Perda de market share;
- Vulnerabilidade de receita.

### Recomendação:

Criar programas de:

- retenção;
